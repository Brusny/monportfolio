## Création des tables

```python
df_saison.createOrReplaceTempView('saison')
df_episode.createOrReplaceTempView('episode')
df_se_nomme.createOrReplaceTempView('se_nomme')
df_pays.createOrReplaceTempView('pays')
df_apparait.createOrReplaceTempView('apparait')
df_personnage.createOrReplaceTempView('personnage')
df_type.createOrReplaceTempView('type')
df_joue.createOrReplaceTempView('joue')
df_acteur.createOrReplaceTempView('acteur')
df_role.createOrReplaceTempView('role')
df_aide.createOrReplaceTempView('aide')
df_cabinet.createOrReplaceTempView('cabinet')
df_travaille.createOrReplaceTempView('travaille')
```

# Requête SQL combinaison de toutes les tables

```python
table1 = spark.sql("""
        SELECT * FROM saison
        INNER JOIN episode
            USING(num_saison)
        INNER JOIN se_nomme
            USING(num_saison, num_episode)
        INNER JOIN(pays)
            USING(id_pays)
        INNER JOIN apparait
            USING(num_saison, num_episode)
        INNER JOIN personnage
            USING(id_personnage)
        INNER JOIN type
            USING(id_type)
        INNER JOIN acteur
            USING(id_personnage)
        LEFT JOIN aide
            USING(num_saison, num_episode, id_personnage)
        LEFT JOIN cabinet
            USING(id_cabinet) 
    """)
table1.show()
table1.createOrReplaceTempView("table1")
```

```python
table2 = spark.sql("""
        SELECT * FROM saison
        INNER JOIN episode
            USING(num_saison)
        INNER JOIN se_nomme
            USING(num_saison, num_episode)
        INNER JOIN pays
            USING(id_pays)
        INNER JOIN joue
            USING(num_saison)
        INNER JOIN personnage
            USING(id_personnage)
        INNER JOIN type
            USING(id_type)
        INNER JOIN acteur
            USING(id_personnage)
        LEFT JOIN role
            USING(id_role)
        LEFT JOIN travaille
            USING(id_personnage)
        LEFT JOIN cabinet
            using(id_cabinet)
        LEFT JOIN aide
            USING(num_saison, num_episode, id_personnage, id_cabinet)
    """)
table2.show()
table2.createOrReplaceTempView("table2")
```

```python
table = spark.sql("""
    SELECT * FROM table1
    FULL JOIN table2
        USING(num_saison, num_episode, id_pays,
        id_personnage, id_acteur, id_type, libelle_type,
        nom_saison, date_debut_saison, date_fin_saison,
        nom_episode, resume_episode, nom_pays,
        nom_personnage, prenom_personnage,nom_acteur, prenom_acteur)
""")
table.show()
```

# Création du fichier CSV

```python
df = table.toPandas()
df.to_csv("scandal.csv", sep=";", index=False, encoding="utf-8")
```
